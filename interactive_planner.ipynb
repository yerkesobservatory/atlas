# Interactive observation planner

Run this notebook with the repository root as its working directory, and with `ipympl` installed (`pip install ipympl`).

1. Run the setup cell.
2. Run either the test-mode cell or the general-mode cell.
3. Drag the bands in the figure to adjust times. Reset restores the starting positions.
4. Run the last cell to get the schedule. It reads the current band positions, so it can be rerun after more adjustments.

```python
%matplotlib widget
import json

import imqueue.schedulers.interactive_planner as ip
```

## Test mode

Edit `TEST_TARGETS` as `{name: exposure in seconds}`. Targets are placed one by one, centered on their time of maximum altitude.

```python
ip.TEST_TARGETS = {
    "M31": 3600,
    "M13": 1800,
    "M57": 1800,
}

NIGHT_DATE = "2026-10-07"   # evening date, local Pacific time

win_start, win_end = ip.compute_night(NIGHT_DATE)
planner = ip.Planner(ip.test_blocks(), win_start, win_end)
```

## General mode

Skip this cell when using test mode. `obs.json` is a JSON list of the observation dictionaries that `general.schedule()` receives (`_id`, `target`, `filters`, `exposure_time`, `exposure_count`, `options.priority`). Starting positions are the times the scheduler produces.

```python
with open("obs.json") as f:
    blocks, starts, win_start, win_end = ip.general_schedule(json.load(f))
planner = ip.Planner(blocks, win_start, win_end, starts)
```

## Final schedule

```python
schedule = planner.schedule()
ip.to_table(schedule).pprint(max_lines=-1, max_width=-1)
```
